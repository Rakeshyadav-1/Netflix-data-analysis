```python
df.drop_duplicates(inplace=True)
```

```python
# Check Duplicated Record
df.duplicated().sum()
```

```python
#  Remove Dirty Values with NaN
df.replace(["N/A","NULL",'',' ',],np.nan,inplace=True)
```

```python
# Extra Spaces Remove
df=df.apply(lambda x: x.str.strip() if x.dtype== "str" else x)
```

```python
# Standardize Text
df["Customer_Name"]=df["Customer_Name"].str.title()
df["Subscription_Plan"]=df["Subscription_Plan"].str.title()
df["Account_Status"]=df["Account_Status"].str.title()
df["Genre"]=df["Genre"].str.title()
df["Country"]=df["Country"].str.title()

```

```python
# Numeric Columns
Numeric_cols = ['Age', 'Watch_Hours', 'Monthly_Fee', 'Movies_Watched',
                'Episodes_Watched', 'Avg_Rating',
                'Monthly_Sessions', 'Download_Count']

for col in Numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
# step remove invalid age
df= df[(df["Age"]>=18) & (df["Age"]<=100)]
```

```python
#Remove negative charges
df=df[df["Watch_Hours"]>0]
df=df[df["Monthly_Fee"]>0]
```

```python
# Missing Values Handle
df['Age']=df['Age'].fillna(df['Age'].mean())
df['Watch_Hours']=df['Watch_Hours'].fillna(df['Watch_Hours'].mean())
df['Monthly_Fee']=df['Monthly_Fee'].fillna(df['Monthly_Fee'].mean())

```

```python
# Fill missing values with Unknown

df['Payment_Method'] = df['Payment_Method'].fillna('Unknown')

df['Device'] = df['Device'].fillna('Unknown')

df['Genre'] = df['Genre'].fillna('Unknown')
df['Country'] = df['Country'].fillna('Unknown')
```

```python
# Feature Engineering

# Customer Revenue
df['Customer_Revenue']=(df['Monthly_Fee']*df['Monthly_Sessions'])
```

```python
# Engagement Score
df['Engagement_Score']=(df['Watch_Hours']+df['Monthly_Sessions']+df['Download_Count'])
```

```python
# Watch Category
bins= [0,20,50,100,np.inf]
label= ['Low','Medium','High','Very High']

df['Watch_Category']=pd.cut(df['Watch_Hours'],bins=bins,labels=label)
```

```python
df['Watch_Category'].value_counts()
```

```python
# Customer Age Group
bins=[17,25,35,50,100]
label=['18-25','26-35','36-50','51+']

df['Age_Group']=pd.cut(df['Age'],bins=bins,labels=label)
```

```python
df['Age_Group'].value_counts()
```

```python
# Account Active Flag
df['Active_Flag']=np.where(df['Account_Status']=='Active',1,0)
```
